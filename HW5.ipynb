```python
import pymc as pm
import numpy as np
import arviz as az
import matplotlib.pyplot as plt

# Generate synthetic normal data
np.random.seed(42)
n = 100
true_theta = 0
true_tau = 1  # Precision (1/variance)
data = np.random.normal(loc=true_theta, scale=1/np.sqrt(true_tau), size=n)

# Define Bayesian model
with pm.Model() as model:
    theta = pm.Beta("theta", alpha=2, beta=2)  # Yet another non-normal prior (Beta)
    tau = pm.Weibull("tau", alpha=2, beta=1)  # Yet another non-gamma prior (Weibull)

    y_obs = pm.Normal("y_obs", mu=theta, sigma=1/pm.math.sqrt(tau), observed=data)

    # Perform MCMC sampling
    idata = pm.sample(2000, return_inferencedata=True)

# Plot diagnostics
az.plot_trace(idata)
plt.tight_layout()
plt.show()

# Print summary statistics
display(az.summary(idata, round_to=2))
```

In slice sampling, the curve that we are sampling beneath is the probability density function (PDF) of the target distribution, denoted as f(x). In the provided code, this function is defined as a Beta distribution with shape parameters (1.5,3), meaning we are sampling from a Beta(1.5, 3) distribution. The goal of slice sampling is to generate samples that follow this distribution by iteratively selecting a region under the curve and drawing new values from it.

The initial value for Gibbs sampling is typically chosen arbitrarily but should be within the domain of the target distribution. In the provided code, the starting value for x is set at 0.25. The Gibbs sampling process proceeds iteratively, updating each variable sequentially. At each step, slice sampling is used to draw a new value for x, ensuring that it remains within the region where f(x)>y. This process is repeated for a specified number of iterations, allowing the Markov chain to converge to the target distribution.

To generate a draw in Gibbs sampling using slice sampling, the process begins by initializing x with an arbitrary value within the domain of the target distribution. Next, a uniform random value y is sampled from the range
[0,f(x)], effectively slicing the probability density function at height y. The algorithm then identifies the horizontal interval where f(x)>y, defining a valid region for sampling. Within this interval, a new x is drawn uniformly to ensure it remains within the sliced region. Finally, this new x value is used as the next sample in the Gibbs chain, and the process is repeated iteratively to generate a sequence of samples that converge to the target distribution.

By incorporating slice sampling within Gibbs sampling, we can efficiently sample from complex distributions where direct sampling is difficult. This approach ensures better exploration of the parameter space compared to Metropolis-Hastings, as it does not require a proposal distribution and dynamically adjusts the sampling interval.
